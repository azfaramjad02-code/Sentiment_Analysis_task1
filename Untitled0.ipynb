```python
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError
import pandas as pd

# Connect to YouTube
youtube = build(
    "youtube",
    "v3",
    developerKey=API_KEY
)

TARGET_COMMENTS = 2000
comments_data = []
seen_comment_ids = set()

# Search for food review videos
search_request = youtube.search().list(
    part="snippet",
    q="food review restaurant review",
    type="video",
    maxResults=50,
    order="relevance"
)

search_response = search_request.execute()

videos = search_response.get("items", [])

print("Food review videos found:", len(videos))

# Collect comments from the videos
for video in videos:

    if len(comments_data) >= TARGET_COMMENTS:
        break

    video_id = video["id"]["videoId"]
    video_title = video["snippet"]["title"]

    print("\nCollecting:", video_title)

    next_page_token = None

    try:
        while len(comments_data) < TARGET_COMMENTS:

            request_params = {
                "part": "snippet",
                "videoId": video_id,
                "maxResults": 100,
                "textFormat": "plainText",
                "order": "relevance"
            }

            if next_page_token:
                request_params["pageToken"] = next_page_token

            response = youtube.commentThreads().list(
                **request_params
            ).execute()

            for item in response.get("items", []):

                comment_info = item["snippet"]["topLevelComment"]
                comment = comment_info["snippet"]
                comment_id = comment_info["id"]

                if comment_id in seen_comment_ids:
                    continue

                seen_comment_ids.add(comment_id)

                comments_data.append({
                    "comment_id": comment_id,
                    "video_id": video_id,
                    "video_title": video_title,
                    "comment": comment["textDisplay"],
                    "published_at": comment["publishedAt"],
                    "like_count": comment["likeCount"]
                })

                if len(comments_data) >= TARGET_COMMENTS:
                    break

            next_page_token = response.get("nextPageToken")

            if not next_page_token:
                break

    except HttpError as error:
        print("Could not collect comments from this video:", error)
        continue

# Create dataset
df = pd.DataFrame(comments_data)

# Save raw comments
df.to_csv(
    "food_review_comments.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nData collection completed!")
print("Total comments collected:", len(df))
print(df.head())
```

```python
import pandas as pd

# Remove missing comments
df = df.dropna(subset=["comment"])

# Remove empty comments
df["comment"] = df["comment"].astype(str).str.strip()

df = df[df["comment"] != ""]

# Remove duplicate comments
df = df.drop_duplicates(subset=["comment"])

# Reset row numbers
df = df.reset_index(drop=True)

print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nFirst five comments:")
print(df[["video_title", "comment"]].head())
```

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

# Download VADER's sentiment dictionary
nltk.download("vader_lexicon")

# Initialize sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Calculate sentiment scores
def analyze_sentiment(text):

    scores = sia.polarity_scores(str(text))

    compound = scores["compound"]

    if compound >= 0.05:
        sentiment = "Positive"

    elif compound <= -0.05:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    return pd.Series({
        "positive_score": scores["pos"],
        "negative_score": scores["neg"],
        "neutral_score": scores["neu"],
        "compound_score": compound,
        "sentiment": sentiment
    })

# Apply sentiment analysis
sentiment_results = df["comment"].apply(analyze_sentiment)

df = pd.concat(
    [df, sentiment_results],
    axis=1
)

print("Sentiment analysis completed!")

print(
    df[["comment", "compound_score", "sentiment"]].head(10)
)
```

```python
import matplotlib.pyplot as plt

sentiment_counts = df["sentiment"].value_counts()

print(sentiment_counts)

sentiment_counts.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Sentiment Analysis of YouTube Food Reviews")
plt.xlabel("Sentiment")
plt.ylabel("Number of Comments")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
